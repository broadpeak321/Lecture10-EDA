## Dataset

### Generate Data using numpy

```python
names = np.char.add('std', np.arange(1,101).astype('str'))   # add characters with number range
names[:10]
```

```python
ages = np.random.randint(18,40,100)
ages[:10]
```

```python
gender = np.random.choice(['Male', 'Female'],100)
gender[:10]
```

```python
# can be done with follwing two methods
# p1 = np.char.add(np.random.randint(10000, 100000, size=100).astype(str), '-')
# p2 = np.char.add(np.random.randint(10000000, 100000000, size=100).astype(str), '-')
# p3 = np.random.randint(0,10,100).astype(str)
# p4 = np.char.add(p1,p2)
# p5 = np.char.add(p4,p3)
# p5

p1 = np.random.randint(10000, 100000, size=100).astype(str)
p2 = np.random.randint(10000000, 100000000, size=100).astype(str)
p3 = np.random.randint(0,10,100).astype(str)

p1 = np.char.add(p1, '-')
p2 = np.char.add(p1, p2)
p3 = np.char.add('-',p3)

cnic = np.char.add(p2,p3)
cnic[:10]
```

```python
# define marks
ml_100 = np.random.randint(20,101,100)
dl_100 = np.random.randint(10,101,100)
nlp_100 = np.random.randint(40,101,100)
eda_60 = np.random.randint(0,61,100)
```

### Introduce Null Values in Data

```python
#randomly make five cnic null
cnic= cnic.astype(object)
cnic[np.random.randint(0,100,5)] = np.nan
```

### Convert it into Dictionary

```python
data = {
    'Name' : names,
    'Age' : ages,
    'Gender': gender,
    'CNIC': cnic,
    'Machine Learning': ml_100,
    'Deep Learning': dl_100,
    'NLP': nlp_100,
    'EDA': eda_60
}
```

### Convert into Data Frame

```python
df = pd.DataFrame(data)
df.head()
```

### Save as CSV file

```python
#index = False   "to remove any unnamed column in your data"
df.to_csv('students_records.csv', index= False)
```

### Feature Engineering

```python
df.head(1)
```

```python
df['Obtained Marks'] = df['Machine Learning'] + df['Deep Learning'] + df['NLP']+df['EDA']
```

```python
df.head(2)
```

```python
df['Percentage'] = (df['Obtained Marks']*100/360).round(2)
```

```python
df.head(1)
```

```python
df['Average Marks'] = (df['Obtained Marks']/4).round(2)
```

```python
df.head(1)
```

```python
df['Status'] = df["Percentage"].apply(lambda x: 'passed' if x>50 else "Failed")
```

```python
df.head(1)
```

```python
def check_status(x):
    if (x>50):
        return "Passed"
    else: 
        return "Failed"
```

```python
df['Status'] = df['Percentage'].apply(check_status)
```
